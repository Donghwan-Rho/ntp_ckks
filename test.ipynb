```python
# Original version

import os
os.environ["PYTHONHASHSEED"] = "42"

import random
random.seed(42)

import numpy as np
np.random.seed(42)


# ==================== cutmax 함수 정의 ====================
def cutmax(
    X: np.ndarray,
    p: int = 9,
    T: int = 6,
    c: float = 1.0,
    noise_scale: float = 0.0,
    top_elets: int = 5,
    eps: float = 1e-12,
    print_topk: int | None = 10,   # None이면 출력 안 함
    return_history: bool = False
):
    def _print_topk(label: str, V: np.ndarray, V_: np.ndarray, X0: np.ndarray, k: int):
        k = min(k, V.size)
        idx_k = np.argpartition(V, -k)[-k:]
        idx_sorted = idx_k[np.argsort(-V[idx_k])]
        print(f"[{label}] Top {k} elements:")
        if label == "Final Z (normalized)":
            for rank, i in enumerate(idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, Y={V[i]:.6g}, Z={V_[i]:.6g}, X={X0[i]:.6g}")
        else:
            for rank, i in enumerate(idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, u={V[i]:.6g}, u^p={V_[i]:.6g}, X={X0[i]:.6g}")
        print("-" * 60)

    X = np.asarray(X, dtype=np.float64).reshape(-1).copy()
    
    # 상위 요소에 노이즈 추가
    if noise_scale > 0:
        top_idx = np.argsort(X)[-top_elets:][::-1]
        X[top_idx] += noise_scale * np.arange(top_elets*2, top_elets, -1, dtype=X.dtype)
        print(f"[noise] Top {top_elets} after noise injection:")
        for rank, i in enumerate(top_idx, 1):
            print(f"  top-{rank}: idx={i}, X={X[i]:.12g}")
    
    # 참조: softmax 결과
    X_shift = X - np.max(X)
    exp_X = np.exp(X_shift)
    softmax_X = exp_X / np.sum(exp_X)
    max_prob = float(np.max(softmax_X))
    max_idx = int(np.argmax(softmax_X))
    print(f"softmax(X) max prob = {max_prob:.6g} (idx={max_idx})")
    
    Y = X.copy()
    hist = [Y.copy()] if return_history else None

    for t in range(1, T + 1):
        mu = Y.mean()
        Y_ = Y - mu
        pos_cnt = np.count_nonzero(Y_ > 0)
        nonpos_cnt = np.count_nonzero(Y_ <= 0)
        print(f"count(Y_ > 0) = {pos_cnt} count(Y_ <= 0) = {nonpos_cnt}")
        s2 = np.mean((Y_) ** 2)
        print(f'mean: {mu} | c: {c} | std: {np.sqrt(s2)}')
        u = (Y_)/(c * np.sqrt(s2 + eps)) + 1.0
        Y = u ** p

        if return_history:
            hist.append(Y.copy())

        if isinstance(print_topk, int) and print_topk > 0:
            _print_topk(f"Iter {t}", u, Y, X, print_topk)

    # 마지막 정규화
    S = Y.sum()
    Z = np.zeros_like(Y)
    if np.abs(S) >= eps:
        Z = Y / S

    if isinstance(print_topk, int) and print_topk > 0:
        _print_topk("Final Z (normalized)", Y, Z, X, print_topk)

    return (Z, hist) if return_history else Z


# ==================== 실행 코드 ====================
if __name__ == "__main__":
    p = 15  # sharpening power
    T = 2   # iterations
    c = 5.0 # normalization constant
    noise_scale = 1
    top_elets = 10  # 상위 몇 개에 노이즈 추가할지
    print_topk = 1  # 상위 몇 개를 출력할지    
    vector_size = 50257
    z_distribution_scale = 3.0  # 정규분포의 scale (std)
    
    # 입력 벡터 생성
    z = np.random.normal(loc=0.0, scale=z_distribution_scale, size=vector_size).astype(np.float64)
    Z = cutmax(z, p=p, T=T, c=c, noise_scale=noise_scale, top_elets=top_elets, 
               print_topk=print_topk, return_history=False)

    # 결과 비교
    approx_argmax = int(np.argmax(Z))
    true_argmax = int(np.argmax(z))
    print(f"approx argmax = {approx_argmax}, true argmax = {true_argmax}")
```

```python
# Lehmer CutMax (T=1)

import os
os.environ["PYTHONHASHSEED"] = "42"

import random
random.seed(42)

import numpy as np
np.random.seed(42)


# ==================== cutmax 함수 정의 ====================
def cutmax(
    X: np.ndarray,
    p0: int = 2,               # pre-sharpening power
    p1: int = 3,               # post-sharpening power
    r: float = 3.0,            # Lehmer order: uses S_r, S_{r+1}
    shift_scale: float = 2.0,  # shift scale for ReLU threshold: μ + t*σ
    noise_scale: float = 0.0,
    top_elets: int = 5,
    eps: float = 1e-20,
    print_topk: int | None = 10,
    return_history: bool = False
):
    """
    Lehmer CutMax (T=1):
    1. X_1 = ReLU(X - μ - t*σ)  (Shifting)
    2. X_2 = X_1^{p0}  (Pre-sharpening)
    3. X_3 = S_r * X_2 - S_{r+1}  (Lehmer mean without normalization)
    4. X_4 = ReLU(X_3^{p1})  (Post-sharpening)
    5. Z = X_4 / sum(X_4)  (L1 normalization)
    """
    def _print_topk(label: str, V: np.ndarray, Vp: np.ndarray, X0: np.ndarray, k: int):
        k = min(k, V.size)
        idx_k = np.argpartition(V, -k)[-k:]
        idx_sorted = idx_k[np.argsort(-V[idx_k])]
        bottom_idx_k = np.argpartition(V, k)[:k]
        bottom_idx_sorted = bottom_idx_k[np.argsort(V[bottom_idx_k])]
        print(f"[{label}] Top {k} elements:")
        if label == "Final Z (normalized)":
            for rank, i in enumerate(idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, X_4={V[i]:.12g}, Z={Vp[i]:.12g}, X={X0[i]:.12g}")
            for rank, i in enumerate(bottom_idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, X_4={V[i]:.12g}, Z={Vp[i]:.12g}, X={X0[i]:.12g}")
        else:
            for rank, i in enumerate(idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, X_3={V[i]:.12g}, X_4={Vp[i]:.12g}, X={X0[i]:.12g}")
            for rank, i in enumerate(bottom_idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, X_3={V[i]:.12g}, X_4={Vp[i]:.12g}, X={X0[i]:.12g}")
        print("-" * 60)

    
    X = np.asarray(X, dtype=np.float64).reshape(-1)
    X_orig = X
    
    # 상위 요소에 노이즈 추가
    if noise_scale > 0:
        top_idx = np.argsort(X)[-top_elets:][::-1]
        X[top_idx] += noise_scale * np.arange(top_elets*2, top_elets, -1, dtype=X.dtype)
        print(f"[noise] Top {top_elets} after noise injection:")
        for rank, i in enumerate(top_idx, 1):
            print(f"  top-{rank}: idx={i}, X={X[i]:.12g}")
    
    # 참조: softmax 결과
    X_shift = X - np.max(X)
    exp_X = np.exp(X_shift)
    softmax_X = exp_X / np.sum(exp_X)
    max_prob = float(np.max(softmax_X))
    max_idx = int(np.argmax(softmax_X))
    print(f"softmax(X) max prob = {max_prob:.6g} (idx={max_idx})")
    
    hist = [X.copy()] if return_history else None

    # Step 1: Compute μ and σ
    mu = X.mean()
    sigma_sq = np.mean((X - mu) ** 2)
    sigma = np.sqrt(sigma_sq)
    threshold = shift_scale * sigma
    
    # Step 2: X_1 = ReLU(X - μ - t*σ) (Shifting)
    X_1 = np.maximum(X - mu - threshold, 0.0)
    exceed_cnt = np.count_nonzero(X > (mu + threshold))
    print(f'count(X > μ+threshold) = {exceed_cnt}')
    
    X_1 /= 15
    
    # Step 3: X_2 = X_1^{p0} (Pre-sharpening)
    X_2 = X_1 ** p0
    
    # Step 4: Compute S_r and S_{r+1}
    if abs(r) < 1e-12:
        S_r = float((X_2 > 0).sum())
    else:
        S_r = float(np.sum(X_2 ** r))
    S_rp1 = float(np.sum(X_2 ** (r + 1.0)))
    
    print(f'μ={mu:.6g}, σ={sigma:.6g}, threshold={threshold:.6g}')
    print(f'r={r}, S_r={S_r:.6g}, S_{{r+1}}={S_rp1:.6g}')
    print(f'X_1 pos={np.count_nonzero(X_1>0)}, nonpos={X_1.size-np.count_nonzero(X_1>0)}')
    
    # Step 5: X_3 = S_r * X_2 - S_{r+1} (Lehmer mean without normalization)
    X_3 = S_r * X_2 - S_rp1
    X_3 /= 10
    
    print(f'X_3 pos={np.count_nonzero(X_3>0)}, nonpos={X_3.size-np.count_nonzero(X_3>0)}')
    
    # Step 6: X_4 = ReLU(X_3^{p1}) (Post-sharpening and eliminating negative values)
    X_4 = np.maximum(X_3 ** p1, 0.0)

    if return_history:
        hist.append(X_4.copy())

    if isinstance(print_topk, int) and print_topk > 0:
        _print_topk(f"After Lehmer", X_3, X_4, X_orig, print_topk)

    # Step 7: Z = X_4 / sum(X_4) (L1 normalization)
    S = X_4.sum()
    print(f'Denominator: {S}')
    Z = np.zeros_like(X_4)
    if abs(S) >= eps:
        Z = X_4 / S

    if isinstance(print_topk, int) and print_topk > 0:
        _print_topk("Final Z (normalized)", X_4, Z, X_orig, print_topk)

    return (Z, hist) if return_history else Z


# ==================== 실행 코드 ====================
if __name__ == "__main__":
    # 데이터 생성 파라미터
    vector_size = 50257
    
    # cutmax 알고리즘 파라미터 & 실행
    z_distribution_scale = 3.0  # 정규분포의 scale (std)
    p0 = 3              # pre-sharpening power
    p1 = 3              # post-sharpening power
    r = 2.0             # Lehmer order
    shift_scale = 4.0   # shift scale (t)
    noise_scale = 1
    top_elets = 10  # 상위 몇 개에 노이즈 추가할지
    print_topk = 2      # 상위 몇 개를 출력할지
    
    # 입력 벡터 생성
    z = np.random.normal(loc=0.0, scale=z_distribution_scale, size=vector_size).astype(np.float64)

    Z = cutmax(z, p0=p0, p1=p1, r=r, shift_scale=shift_scale, noise_scale=noise_scale, 
               top_elets=top_elets, print_topk=print_topk, return_history=False)
    
    print(f'(noise_scale, max(Z)): ({noise_scale}, {max(Z)})')

    # 결과 비교
    approx_argmax = int(np.argmax(Z))
    true_argmax = int(np.argmax(z))
    print(f"approx argmax = {approx_argmax}, true argmax = {true_argmax}")
```

```python
# 1-relu version (Lehmer-q generalized)

import os
os.environ["PYTHONHASHSEED"] = "42"

import random
random.seed(42)

import numpy as np
np.random.seed(42)

def cutmax(
    X: np.ndarray,
    p: int = 8,           # sharpening power
    T: int = 6,           # iterations
    c: float = 1.0,
    q: float = 2.0,       # <-- Lehmer order (q): uses S_{q-1}, S_q
    eps: float = 1e-12,
    print_topk: int | None = 10,
    return_history: bool = False
):
    def _print_topk(label: str, V: np.ndarray, V_: np.ndarray, X0: np.ndarray, k: int):
        k = min(k, V.size)
        idx_k = np.argpartition(V, -k)[-k:]
        idx_sorted = idx_k[np.argsort(-V[idx_k])]
        print(f"[{label}] Top {k} elements:")
        if label == "Final Z (normalized)":
            for rank, i in enumerate(idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, Y={V[i]:.6g}, Z={V_[i]:.6g}, X={X0[i]:.6g}")
        else:
            for rank, i in enumerate(idx_sorted, 1):
                print(f"  {rank:2d}) idx={i:6d}, u={V[i]:.6g}, u^p={V_[i]:.6g}, X={X0[i]:.6g}")
        print("-" * 60)

    X = np.asarray(X, dtype=np.float64).reshape(-1)
    Y = X.copy()
    hist = [Y.copy()] if return_history else None

    for t in range(1, T + 1):
        # ===== 1-ReLU 스큐 표준화 =====
        tau = Y.mean()
        a = Y - tau
        pos_cnt = np.count_nonzero(a > 0)
        nonpos_cnt = np.count_nonzero(a <= 0)
        print(f"count(a > 0)  = {pos_cnt} | count(a <= 0) = {nonpos_cnt}")

        a_plus = np.maximum(a, 0.0)  # ReLU: 이터당 1회

        # ----- 일반 Lehmer mean용 S_{q-1}, S_q -----
        # q=1인 특수케이스에서 0^0을 피하기 위해 양수 항만 1로 세서 S_{0}를 정의
        if abs(q - 1.0) < 1e-12:
            S_qm1 = float((a_plus > 0).sum())  # S_{0} ≈ # of positive terms
        else:
            S_qm1 = np.sum(a_plus ** (q - 1.0))

        S_q = np.sum(a_plus ** q)

        print(f"mean: {tau}")
        print(f"S_(q-1): {S_qm1}  |  S_q: {S_q}")

        # u = a_plus * S_{q-1} / (S_q + eps)  == a_plus / (delta_q + eps/S_{q-1})
        denom = S_q + eps
        scale = S_qm1 / denom if denom > 0 else 0.0
        u = a_plus * scale

        # 샤프닝
        Y = u ** p

        if return_history:
            hist.append(Y.copy())

        if isinstance(print_topk, int) and print_topk > 0:
            _print_topk(f"Iter {t}", u, Y, X, print_topk)

    # 마지막 정규화
    S = Y.sum()
    Z = np.zeros_like(Y)
    if np.abs(S) >= eps:
        Z = Y / S

    if isinstance(print_topk, int) and print_topk > 0:
        _print_topk("Final Z (normalized)", Y, Z, X, print_topk)

    return (Z, hist) if return_history else Z


# ---------------------- 사용 예시 ----------------------
z = np.random.uniform(-10, 10, 50257).astype(np.float64)

p, T, c = 3, 3, 1.0
# q=1.0 -> L_1 = S2/S1 (contraharmonic), q=2.0 -> L_2 = S3/S2, ...
Z = cutmax(z, p=p, T=T, c=c, q=2.0, print_topk=10, return_history=False)

approx_argmax = int(np.argmax(Z))
true_argmax   = int(np.argmax(z))
print(f"approx argmax = {approx_argmax}, true argmax = {true_argmax}")
```

```python
# 2-relu version

import os
os.environ["PYTHONHASHSEED"] = "42"

import random
random.seed(42)

import numpy as np
np.random.seed(42)

def cutmax(
    X: np.ndarray,
    p: int = 9,
    T: int = 6,
    c: float = 1.0,
    eps: float = 1e-12,
    print_topk: int | None = 10,   # None이면 출력 안 함
    return_history: bool = False
):
    def _print_topk(label: str, V: np.ndarray, X0: np.ndarray, k: int):
        k = min(k, V.size)
        idx_k = np.argpartition(V, -k)[-k:]
        idx_sorted = idx_k[np.argsort(-V[idx_k])]
        print(f"[{label}] Top {k} elements:")
        for rank, i in enumerate(idx_sorted, 1):
            print(f"  {rank:2d}) idx={i:6d}, val={V[i]:.6g}, X={X0[i]:.6g}")
        print("-" * 60)

    X = np.asarray(X, dtype=np.float64).reshape(-1)
    Y = X.copy()
    hist = [Y.copy()] if return_history else None

    # -----(스큐 표준화에 쓰는 내부 하이퍼파라미터)-----
    k_clip = 3.0     # 가중치 클리핑 상한: u = clip(Y-mean, [-1, k_clip])
    q_w   = 4        # 가중치 지수(= 2^m 권장; 여기선 4)
    gamma = 1e-12     # 꼬리 스케일 바닥(0 division 방지 및 온도)

    for t in range(1, T + 1):
        # ===== ReLU-스큐 표준화 영역 (교체된 부분) =====
        tau = Y.mean()
        a = Y - tau
        # 가중치: 평균 미만은 0, 이상은 (1..(k+1)^q_w)
        # a_clip = np.minimum(np.maximum(a, -1.0), k_clip)
        # w = (a_clip + 1.0) ** q_w
        a_clip = np.maximum(a, 0)
        w = a_clip

        S = w.sum() + eps                     # 분모(가중치 합)
        mu_plus = (w * a_clip).sum() / S           # ReLU-가중 평균(스큐 평균)
        print(f'mu_plus: {mu_plus}')
        r_pos = np.maximum(Y - mu_plus, 0.0)  # 양의 꼬리만
        delta_plus = (w * r_pos).sum() / S    # 양의 꼬리의 전형적 폭

        # 스큐 표준화된 양수부 (u >= 0). 기존 c는 스케일 조절로 반영.
        u = r_pos / (gamma + delta_plus)      # √/분산 없이 one-sided scale
        # 기존 CutMax의 (+1)^p 샤프닝 형식을 유지
        # Y = (u / max(c, eps) + 1.0) ** p
        # Y = u ** p
        Y = u
        # ==============================================

        if return_history:
            hist.append(Y.copy())

        if isinstance(print_topk, int) and print_topk > 0:
            _print_topk(f"Iter {t}", Y, X, print_topk)

    # 마지막 정규화
    S = Y.sum()
    Z = np.zeros_like(Y)
    if np.abs(S) >= eps:
        Z = Y / S

    # 정규화 후 Top-10 출력
    if isinstance(print_topk, int) and print_topk > 0:
        _print_topk("Final Z (normalized)", Z, X, print_topk)

    return (Z, hist) if return_history else Z


# ---------------------- 사용 예시 ----------------------
z = np.random.uniform(-10, 10, 32000).astype(np.float64)
# z[np.argmax(z)] += 0.01  # 옵션

p, T, c = 15, 10, 1.0
Z = cutmax(z, p=p, T=T, c=c, print_topk=5, return_history=False)

approx_argmax = int(np.argmax(Z))
true_argmax   = int(np.argmax(z))
print(f"approx argmax = {approx_argmax}, true argmax = {true_argmax}")
```

```python
# GK version (underflow)

import os
os.environ["PYTHONHASHSEED"] = "42"

import random
random.seed(42)

import numpy as np
np.random.seed(42)

def cutmax(
    X: np.ndarray,
    # --- 핵심 하이퍼파라미터 ---
    p: int = 4,             # exp 근사 깊이: e(x) ≈ (1 + x/2^p)^(2^p)
    T: int = 6,             # 반복 횟수
    c: float = 1000000.0,         # 샤프닝 후 스케일(폭발 방지)
    k: float = 1000000.0,      # 램프 폭 (전이 폭은 ~1/k)
    lb: float = -0.0,       # 평균 이하 영역을 수축시킬 하한(상수)
    pow_out: float = 1.0,   # exp 이후 추가 샤프닝 지수(1.0이면 비활성)
    # --- 기타 ---
    eps: float = 1e-12,
    print_topk: int | None = 10,   # None이면 출력 안 함
    return_history: bool = False
):
    """
    Iteration (ReLU 1회 사용):
      tau  = mean(Y)
      a    = Y - tau
      r1   = relu(a)
      r2   = relu(a - 1/k)
      m    = k * (r1 - r2)   # ramp in [0,1]
      x_t  = r1 + lb * (1 - m)
      e_p  = (1 + x_t / 2^p)^(2^p)   # exp 근사
      U    = ((e_p - 1) * m) / c
      Y    = U ** pow_out
    마지막에 한 번 정규화(Z = Y / sum Y).
    주의: exp 근사에서 base = 1 + x_t/2^p > 0 이어야 하므로 lb > -2^p 권장.
    """

    def _print_topk(label: str, V: np.ndarray, X0: np.ndarray, k_: int):
        k_ = min(k_, V.size)
        idx_k = np.argpartition(V, -k_)[-k_:]
        idx_sorted = idx_k[np.argsort(-V[idx_k])]
        print(f"[{label}] Top {k_} elements:")
        for rank, i in enumerate(idx_sorted, 1):
            print(f"  {rank:2d}) idx={i:6d}, val={V[i]:.6g}, X={X0[i]:.6g}")
        print("-" * 60)

    X = np.asarray(X, dtype=np.float64).reshape(-1)
    Y = X.copy()
    hist = [Y.copy()] if return_history else None

    # 고정 상수
    inv_pow = 1.0 / (2.0 ** p)     # 1 / 2^p
    exp_power = 2 ** p             # 2^p

    for t in range(1, T + 1):
        # 1) 센터링
        if t == 1:
            Y /= (32000*20)
            
        tau = Y.mean()
        a = Y - tau
                    
        r = np.maximum(a, 0.0)
        s = np.sum(r)
        x = (-1/2)*(Y - s)**2
        gk = np.exp(x)
        print(f'gk: {gk}')
        
        gk /= 32000
        
        Y = gk

        if return_history:
            hist.append(Y.copy())
        if isinstance(print_topk, int) and print_topk > 0:
            _print_topk(f"Iter {t}", Y, X, print_topk)

    # 마지막 정규화 (합 1)
    S = Y.sum()
    Z = np.zeros_like(Y)
    if np.abs(S) >= eps:
        Z = Y / S

    # 정규화 후 Top-k 출력
    if isinstance(print_topk, int) and print_topk > 0:
        _print_topk("Final Z (normalized)", Z, X, print_topk)

    return (Z, hist) if return_history else Z


# ---------------------- 사용 예시 ----------------------
z = np.random.uniform(-10, 10, 32000).astype(np.float64)
# z[np.argmax(z)] += 0.01  # 필요시 top-1 버프 테스트

# p: exp 근사 깊이(깊을수록 exp에 가까움), k: 램프 폭(전이 폭 ~ 1/k),
# lb: 평균 이하 수축 강도(너무 작게 두지 말 것: lb > -2^p 권장), c: 스케일
p, T, c = 8, 5, 5.0
Z = cutmax(
    z, p=p, T=T, c=c,
    k=1500.0,     # 전이 폭을 더 날카롭게
    lb=-64.0,     # p=8이면 -2^p=-256 보다 크게
    pow_out=1.0,  # 추가 샤프닝 비활성(원하면 1.5~3.0 등으로)
    print_topk=10,
    return_history=False
)

approx_argmax = int(np.argmax(Z))
true_argmax   = int(np.argmax(z))
print(f"approx argmax = {approx_argmax}, true argmax = {true_argmax}")
```

```python
import numpy as np

# 예시 벡터 (Lehmer 평균은 보통 비음수에서 정의)
x = np.array([0.5, 2.0, 3.5, 9.0, 10.0, 10.0], dtype=float)

# 산술평균
arith_mean = x.mean()

def lehmer_mean(x: np.ndarray, r: float = 1.0, eps: float = 1e-12) -> float:
    """
    L_r = (sum x^{r+1}) / (sum x^r)
    r=0이면 L_0는 산술평균과 같음.
    비음수 입력을 가정.
    """
    x = np.asarray(x, dtype=float)
    if abs(r) < 1e-12:   # r = 0 -> S0 = sum x^0 = n
        Sr = float(x.size)
    else:
        Sr = float(np.sum(x ** r))
    Srp1 = float(np.sum(x ** (r + 1.0)))
    return Srp1 / (Sr + eps)

print("x:", x)
print(f"Arithmetic mean     = {arith_mean:.6f}")
print(f"Lehmer mean L0      = {lehmer_mean(x, r=0):.6f}  (equals arithmetic mean)")
print(f"Lehmer mean L1      = {lehmer_mean(x, r=1):.6f}  (contraharmonic)")
print(f"Lehmer mean L2      = {lehmer_mean(x, r=2):.6f}")
print(f"Lehmer mean L3      = {lehmer_mean(x, r=3):.6f}")
print(f"Lehmer mean L4      = {lehmer_mean(x, r=4):.6f}")
```
